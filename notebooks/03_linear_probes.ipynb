```python
import sys
sys.path.insert(0, '..')
from src.model import load_model
from src.probes import build_probe_dataset, train_test_split_by_prompt, standardize
from src.probe_model import train_all_layers, random_baseline_accuracy, probe_accuracy_by_layer
from src.io import load_probes

model = load_model('gemma-2-2b')
df = load_probes().head(10)
prompts_by_language = {
    'sinhala': list(zip(df['id'], df['sinhala'])),
    'english': list(zip(df['id'], df['english'])),
}
dataset = build_probe_dataset(model, prompts_by_language)
train_idx, test_idx = train_test_split_by_prompt(dataset['prompt_ids'], dataset['y'])
results = train_all_layers(dataset['X'], dataset['y'], train_idx, test_idx)
print(probe_accuracy_by_layer(results))
print('baseline:', random_baseline_accuracy(dataset['y']))
```
